```python
import pandas as pd
import numpy as np
```

```python
total_cuts = 0
total_0_1 = 0

for i in range(100):

    path = f'../data/simulation/coord_diff/test6_multi/transition_times_with_edges_{i}.csv'

    df = pd.read_csv(path, header=None, skiprows=1)

    total_0_1 += sum(df[4])
    total_cuts += df.shape[0]

total_cuts, total_0_1, total_0_1 / total_cuts
```
