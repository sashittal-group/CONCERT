```python
tumors_per_metfamily = (
    df.groupby("MetFamily")["Tumor"]
      .nunique()
      .reset_index(name="n_tumors")
)

tumors_per_metfamily[tumors_per_metfamily['n_tumors'] > 1]
```

```python
tumor_list = pd.read_csv("../data/KPTracer-Data/trees/tumor_list.txt")
tumor_list_all = tumor_list[tumor_list['Tumor'].str.contains('All')]
tumor_list_all
```

```python
tumors_per_mouse = (
    df.groupby("Mouse")
      .agg(
          n_anatomical_sites=("Tumor", "nunique"),
          n_cells=("Tumor", "size")
      )
      .reset_index()
)

tumors_per_mouse
```
